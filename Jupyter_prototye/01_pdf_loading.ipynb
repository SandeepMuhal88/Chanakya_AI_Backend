```python
import os
from dotenv import load_dotenv

load_dotenv()

HF_TOKEN = os.getenv("HF_TOKEN")

if not HF_TOKEN:
    raise ValueError("HF_TOKEN is missing from .env")

print("Hugging Face token loaded successfully.")
```

Hugging face API Test


```python
from huggingface_hub import InferenceClient

client = InferenceClient(
    api_key=HF_TOKEN
)

print("Hugging Face Inference Client initialized successfully.")
```

```python
import os
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

llm = HuggingFaceEndpoint(
    repo_id="Qwen/Qwen2.5-7B-Instruct",   # -1M hata diya — ye providers pe available hai
    task="text-generation",
    provider="auto",                       # HF khud sahi provider select karega
    max_new_tokens=512,
    huggingfacehub_api_token=os.environ["HF_TOKEN"],
)
model = ChatHuggingFace(llm=llm)
question = "What is the capital of France?"
response = model.invoke(question)
print(response.content)
```
